```python
import pandas as pd
```

importation du projet de la mel pour filtrer

```python
df_mel = pd.read_csv("/data/Projet_reconversion/Projet et cours Data Analyst/Cours Le Wagon/projet_crise_logement/fichier_mel_corrigé.csv")
df_mel.columns
```

```python
code_postal = df_mel["code_postal"].tolist()
```

creation d une liste des code postaux de la mel pour selectionner les meme dans le fichier FD_LOGEMTZB_2022

```python
chunksize = 100_000
dfs = []

for chunk in pd.read_csv(
    "FD_LOGEMTZB_2022.csv",
    chunksize=chunksize,
    sep=";"
):
    filtered = chunk[chunk["COMMUNE"].astype(str).str.startswith("59")]
    dfs.append(filtered)

df = pd.concat(dfs)
df.head()
```

Le fichier était trop volumineux donc j'ai du le partionner en 100000 lignes, j ai ensuite recherché les lignes qui commence par 59 pour faire un premier filtre 

```python
df.shape
```

affichage de la shape du dataframe pour évaluer la véracitée des données 

```python
df = df[df["COMMUNE"].isin(code_postal)]
df.shape
```

filtre final pour les communes de la mel uniquement 

```python
df.to_csv("Caracteristique_logement_mel_2022.csv", index=False)
```
