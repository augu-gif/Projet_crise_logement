```python
import pandas as pd

repartition_statut_occupation = pd.read_csv("C:/Users/gromm/Downloads/Parc de logements.xlsx - Figure 5.csv")
display(repartition_statut_occupation)
```

```python
# Renamed columns

repartition_statut_occupation.columns = ["annee", "ensemble_proprio", "proprio_sans_remboursement", 
                                         "proprio_accedant_propriete", "ensemble_locataires", "bailleurs_publics", "bailleurs_prives",
                                         "loges_gratuitement"]

# Removed the 2 first and the 5 last rows

repartition_statut_occupation = repartition_statut_occupation.drop(repartition_statut_occupation.index[:2])
repartition_statut_occupation = repartition_statut_occupation.drop(repartition_statut_occupation.index[-5:])

display(repartition_statut_occupation)
```
