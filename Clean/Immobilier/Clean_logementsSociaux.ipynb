```python
import pandas as pd

logementsSociaux = pd.read_csv("C:/Users/gromm/Downloads/LogementsSociaux_France_et_Communes - Data.csv")
display(logementsSociaux)
```

```python
# Renamed columns

logementsSociaux.columns = ["code_geo", "lib_geo", "année", "nb_rpls"]

# Removed the 4 first rows of the data

logementsSociaux = logementsSociaux.drop(logementsSociaux.index[:4])
display(logementsSociaux)
```

```python
# Kept the departments from "Nord"

logementsSociaux = logementsSociaux[logementsSociaux['code_geo'].str.startswith('59')]
display(logementsSociaux)
```

```python
# Filled missing years

logementsSociaux['année'] = logementsSociaux["année"].fillna("2023")
display(logementsSociaux)
```

```python
# Export

logementsSociaux.to_csv("LogementsSociaux.csv", index = False)
```
