```python
import pandas as pd

logements_insalubres = pd.read_csv("C:/Users/gromm/Downloads/LogementsInsalubres_France - LogementsInsalubres_France.csv")
display(logements_insalubres)
```

```python
# Set the type of 2 rows from float to int because that didn't make sense

logements_insalubres['année d’entrée dans l’observatoire'] = logements_insalubres['année d’entrée dans l’observatoire'].astype(int)
logements_insalubres['mois d’entrée dans l’observatoire'] = logements_insalubres['mois d’entrée dans l’observatoire'].astype(int)
display(logements_insalubres)
```
