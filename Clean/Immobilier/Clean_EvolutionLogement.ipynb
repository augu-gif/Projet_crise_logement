```python
import pandas as pd

evol_nb_logements = pd.read_csv("C:/Users/gromm/Downloads/Parc de logements.xlsx - Figure 2.csv")
display(evol_nb_logements)
```

```python
# Renamed columns

evol_nb_logements.columns = ["annee", "ensemble_logements", "residences_principales", "residences_secondaires", "logement_vacants", "population"]

# Removed the 4 last and the 2 first rows because they were describing what the data does but wasn't actionnable

evol_nb_logements = evol_nb_logements.drop(evol_nb_logements.index[-4:])
evol_nb_logements = evol_nb_logements.drop(evol_nb_logements.index[:2])

display(evol_nb_logements)
```

```python
# Removed the "r" and the "p" on years

evol_nb_logements["annee"] = evol_nb_logements["annee"].str.replace("r", '')
evol_nb_logements["annee"] = evol_nb_logements["annee"].str.replace("p", '')
display(evol_nb_logements)
```
