filtre par code postal de la mel

```python
df_mel = pd.read_excel("/content/drive/MyDrive/Projet_crise_logement/Données_nettoyées/Clef_communes/INSEE_CODEPOSTAL.xlsx")
```

```python
code_departement = df_mel["CODE_POSTAL"].tolist()
```

```python
df = df_concat[df_concat["Code postal"].isin(code_departement)]
```

utilisation du code postal pour prendre les villes de la mel

```python
df
```

```python
df.dtypes
```

```python
df = df[df["Type local"].isin(["Maison", "Appartement"])]

df["Date mutation"] = pd.to_datetime(df["Date mutation"], dayfirst=True)

df["Code postal"] = df["Code postal"].astype(int).astype(str)

df["Valeur fonciere"] = (df["Valeur fonciere"].str.replace(",", ".", regex=False))

df["Valeur fonciere"] = pd.to_numeric(df["Valeur fonciere"], errors="coerce")

df["prix_m2"] = (df["Valeur fonciere"] / df["Surface reelle bati"]).round(2)

df_mel["CODE_POSTAL"] = df_mel["CODE_POSTAL"].astype(str)
```

transformation du type des colonnes pour power BI

```python
df.dtypes
```

```python
df.isna().sum()
```

```python
df = df.dropna()
```

suppression des valeurs null

```python
df["Valeur fonciere"].min()
```

```python
df = df[
    (df["Surface reelle bati"] >= 10) &
    (df["Surface reelle bati"] <= 500) &
    (df["Valeur fonciere"] >= 10000) &
    (df["Valeur fonciere"] <= 2000000)
]
```

```python
df["prix_m2"].min()
```

```python
df = df[(df["prix_m2"] >= 300) & (df["prix_m2"] <= 10000)]
```

```python
df.shape
```

```python
df["Commune"].nunique()
```

suppression des valeurs abérrantes

```python
communes_a_supprimer = [
    "BACHY",
    "BOURGHELLES",
    "CAMPHIN-EN-PEVELE",
    "COBRIEUX",
    "CYSOING",
    "TRESSIN",
    'LOUVIL',
    "WANNEHAIN"
]

df = df[~df["Commune"].isin(communes_a_supprimer)]
```

```python
df["Commune"] = df["Commune"].replace({
    "FACHES THUMESNIL": "FACHES-THUMESNIL",
    "MONS EN BAROEUL": "MONS-EN-BAROEUL",

})
```

```python
df = df.merge(
    df_mel,
    left_on="Code postal",
    right_on="CODE_POSTAL",
    how="left"
)
```

```python
df = df.drop(columns=["INSEE", "LIBVILLE", "CODE_POSTAL", "N_INSEE"])
```
