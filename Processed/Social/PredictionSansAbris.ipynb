```python
import pandas as pd

sdf = pd.read_csv("C:/Users/gromm/Downloads/SansAbris_toWork.csv")

display(sdf)
```

```python
from sklearn.linear_model import LinearRegression

# données connues
train = sdf.dropna()

X = train[["Année"]]
y = train["sans_abris"]

model = LinearRegression()
model.fit(X, y)

# lignes à prédire
missing = sdf[sdf["sans_abris"].isna()]

pred = model.predict(missing[["Année"]])

sdf.loc[sdf["sans_abris"].isna(), "sans_abris"] = pred

display(sdf)
```

```python
# Export 

sdf.to_csv("SansAbris.csv", index = False)
```
